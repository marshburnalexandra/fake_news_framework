## 1. **DESCRIPTION OF THE PROJECT**

This project implements a complete Fake News Detection Framework designed to process raw text, clean and normalize it, convert it into numerical representations, train machine-learning models, evaluate their performance, and use the trained models to classify new unseen news article as either _real_ or _fake_.  

Unlike simple notebooks that mix preprocessing, feature extraction, and modeling inside a single file, this project follows a modular package-based architecture inspired by real-world production machine learning pipelines. Each part of the project handles a single responsibility, making the system reusable, organized, and easy to extend with additional models or preprocessing options in the future.  

The system includes the following major components:  

- A **Preprocessing Module** to clean text using steps such as lowercasing, tokenization, URL removal, stopword filtering, and stemming.
- A **Feature Extraction Module** that transforms processed text into TF-IDF feature vectors.
- A **Model Module** that provides three machine learning algorithms: Logistic Regression, Naive Bayes, and Support Vector Machine.
- A **Pipeline Class** that connects preprocessing, feature extraction, and modeling into one automated workflow.
- A **Utility Module** for loading datasets and previewing samples.
- A **Dataset** consisting of nearly 7,000 labeled examples of real and fake news.

This project demonstrates how a real machine learning system is built behind the scenes - organized, separated by functionality and scalable.

## 3. **How Each File Works and How to Use Them**

**a. `__init__.py`**  
This file makes the package easier to import by exposing all major classes at the package level.  
Because of this file, the user can simple write:  
```python
from fakenews import FakeNewsPipeline, LogisticNewsModel, load_dataset
```
Instead of importing each file separately.  
It improves usability, cleaniness and abstraction.  

**b. `preprocessing.py`**  
This module contains the `Preprocessor` class, which cleans raw text before sending it to the machine learning model.  

It performs:  

- Lowercasing
- URL removal
- Removal of numbers and punctuation
- Stopword filtering
- Word stemming

The preprocessing ensures the model receives normalized and meaningful input. It can be configured by enabling or disabling stopword removal and stemming.  

You use it like this:  
```python
from fakenews import Preprocessor
p = Preprocessor()
cleaned_text = p.clean("THIS is Just a SaMpLe!! https://eskemburdosh.com")
```

**c. `features.py`**  
This file defines the `FeatureExtractor` class, which uses TF-IDF to convert cleaned text into numerical vectors.  

It includes:

- Vocabulary generation
- TF-IDF transformation
- Support for training and inference  
  
It integrates directly with the Preprocessor so users never need to manually clean text before extracting features.  

**d. `models.py`**  
This module contains the models used for training:  

- Logistic Regression
- Multinomial Naive Bayes
- Support Vector Machine

All models follow a shared abstract base class for a consistent interface:  

- `train(X, y)`
- `predict_single(x)`
- `evaluate(X, y)`

This file makes the system expandable - additional models can be added easily.  

**e. `pipeline.py`**  
This is the core system that connects:  

**1.** Preprocessing  
**2.** Feature extraction  
**3.** Model training  
**4.** Prediction  
**5.** Evaluation  

The user only needs to interact with this class.  
It hides all the complexity behind a simple interface.  

Example usage:
```python
pipeline.fit(texts, labels)
pipeline.predict("Some idk news article...")
pipeline.evalute(test_texts, test_labels)
```

**f. `utils.py`**  
This module contains helper functions:  

- `load_dataset(path)`
  - Automatically detects CSV, JSON, or TXT formats
  - Detects which columns contain text or labels
  - Cleans missing values
- `preview(df)`
  - Prints the top rows of the dataset

This ensures dataset loading always behaves predictably.  

### **OOP Concepts in the Fake News Detection Framework**

**1. Classes**  
In this project, nearly every functional component is implemented as a class. A class is a blueprint that defines both data (attributes) and behavior (methods). By structuring the project around classes, each major responsibility is isolated into its own component.  

For example:  

- The `Preprocessor` class handles cleaning and normalizing text.
- The `FeatureExtractor` class transforms cleaned text into TF-IDF vectors.
- The model classes (`LogisticNewsModel`, `NaiveBayesNewsModel`, `SVMNewsModel`) handle training, prediction, and evaluation.
- The `FakeNewsPipeline` class orchestrates the entire workflow by connecting preprocessing, feature extration, and modeling.
- Utility functions reside in `utils.py`, but the core functionality of the system is object-oriented and class-driven.

Using classes allows each component to be reused, instantiated multiple times with different configurations, and easily replaced without modifying the entire system. This is the foundation of our project's modular design. 

**2. Inheritance**  
Inheritance is one of the most important OOP concepts implemented in the project. It allows a child class to inherit methods and behavior from a parent class while adding or overriding functionality. 

In this system, inheritance appears most clearly in the machine learning models. 

All model classes inherit from a shared abstract base class:
```python
class BaseNewsModel(ABC):
```
This base class defines the required interface for all news-classification models. It declares abstract methods such as:

- `train()`
- `predict_single()`
- `evaluate()`

The child model classes:  

- `LogisticNewsModel`
- `NaiveBayesNewsModel`
- `SVMNewsModel`

all inherit from the `BaseNewsModel` and implement their own versions of these required methods.  

This means every model shares the same interface while still being free to use different machine-learning algorithms internally. Because of inheritance, the pipeline does not need to know what model it is working with - it simply expects the model to obey the structure of the base class.  

**3. Encapsulation**  
Encapsulation refers to the bundling of data and the methods that operate on the data inside a single unit. It also involves controlling access so components cannot interfere with each other's internal state unless explicitly allowed.  

Our project uses encapsulation in several ways:

The `Preprocessor` class hides all preprocessing rules inside the class. The user does not need to know how URLs are removed, how punctuation is filtered, or how stemming works. They simple call:
```python
cleaned = preprocessor.clean(text)
```
and the class handles all internal processes.  

The `FeatureExtractor` class encapsulates:  

- the TF-IDF vectorizer
- the vocabulary training
- the transformation of new data

Users do not interact with the vectorizer directly, they only communicate with the FeatureExtractor.  

The model classes encapsulate the actual scikit-learn model objects inside attributes like `self.model`. The user cannot modify the internals unless the class allows it.  

The `FakeNewsPipeline` class encapsulates the entire machine-learning workflow. It contains a preprocessor, feature extractor, and model instance internally. Users cannot directly change the TF-IDF vectorizer, nor do they need to call preprocessing manually. The pipeline exposes only the essential methods:  

- `fit()`
- `predict()`
- `evaluate()`

All internal operations are hidden behind these public methods. This is exactly what encapsulation is designed for: controlling access, protecting internal logic, and providing a clean simple interface. 

**4. Polymorphism**
Polymorphism means "many forms". It allows multiple classes to share the same method names but implement different behavior internally. This project uses polymorphism extensively in the model system.  

The base class `BaseNewsModel` declares three required behaviors:

- `train()`
- `predict_single()`
- `evaluate()`

Each child model class implements these methods in its own way:

- Logistic Regression uses a logistic function for classification.
- Naive Bayes uses probabilistic estimation.
- SVM uses hyperplanes for separation.

Even though the implementations differ, the pipeline interacts with all models using the exact smae method names. This is only possible because of polymorphism.

For example, calling:
```python
self.model.train(X, y)
```
inside the pipeline will work whether the model is Logistic Regression, Naive Bayes, SVM, or even a new model added in the future.  

This allows the user to switch models instantly:
```python
pipeline = FakeNewsPipeline(model=LogisticNewsModel())
```
without changing a single line of pipeline code.
The system becomes flexible, extendable, and easy to upgrade - all due to polymorphism/

## 6. **How to Call the Functions in the Project** 
  
### **DEMONSTRATION CODE**  
---

### **Cell 1 - Import Everything**  

````python
from fakenews import (
    FakeNewsPipeline,
    Preprocessor,
    FeatureExtractor,
    LogisticNewsModel,
    NaiveBayesNewsModel,
    SVMNewsModel,
    load_dataset,
    preview
)
````

### **Cell 2 - Load the Dataset**

````python
df = load_dataset("news.csv")
preview(df)
````

### **Cell 3 - Split into Training and Testing**

````python
from sklearn.model_selection import train_test_split

texts = df["text"].tolist()
labels = df["label"].tolist()

X_train, X_test, y_train, y_test = train_test_split(
    texts, labels, test_size=0.2, random_state=42
)
````

### **Cell 4 - Create Pipeline with Selected Model**  

````python
model = LogisticNewsModel()
pipeline = FakeNewsPipeline(model=model)
````

### **Cell 5 - Train the Model**

````python
pipeline.fit(X_train, y_train)
````

### **Cell 6 - Evaluate**

````python
pipeline.evaluate(X_test, y_test)
````

### **Cell 7 - Predict New Input**

````python
pipeline.predict("Breaking news: Government confirms aliens have arrived on Earth!")
````
